```python
# Faster than list
import numpy as np
import time

L1 = range(1, 100000)
L2 = range(1, 100000)

A1 = np.array(L1)
A2 = np.array(L2)

# Operation to be performed on the list
start = time.time()

z = zip(L1, L2)
result1 = [x + y for x, y in z]

end = time.time()

print("List time:", (end - start) * 1000, "ms")


# Operation to be performed on NumPy array
start1 = time.time()

result2 = A1 + A2

end1 = time.time()

print("NumPy time:", (end1 - start1) * 1000, "ms")
```

```python
# create  a matrix of 3 x 4 of random number and find its maximun and minimum
import numpy as np

# Create a 3 x 4 matrix of random numbers
matrix = np.random.randint(1, 100, size=(3, 4))

print("Matrix:")
print(matrix)

# Find maximum and minimum
maximum = np.max(matrix)
minimum = np.min(matrix)

print("Maximum:", maximum)
print("Minimum:", minimum)
```
