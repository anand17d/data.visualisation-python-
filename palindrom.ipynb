```python
n = int(input("Enter a number: "))

original = n
reverse = 0

while n > 0:
    digit = n % 10
    reverse = reverse * 10 + digit
    n = n // 10

if original == reverse:
    print("Palindrome")
else:
    print("Not Palindrome")
```

```python
def factorial (num):
    if num == 0:
        return 1
    else:
        return num*factorial(num-1)
print(factorial(6))
```
