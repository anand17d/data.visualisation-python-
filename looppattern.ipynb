```python
import random

secret = random.randint(1, 20)

attempt = 0

while True:
    guess = int(input("Enter the number: "))

    attempt += 1

    if guess == secret:
        print(f"You guessed it in {attempt} attempts")
        break

    elif guess < secret:
        print("Too low")

    elif guess > secret:
        print("Too high")
```

```python
print("Think of a number between 1 and 100.")
print("I will try to guess it!")

low = 1
high = 100
attempts = 0

while low <= high:
    attempts += 1
    guess = (low + high) // 2
    print(f"My guess is: {guess}")
    
    feedback = input("Is it correct (c), too high (h), or too low (l)? ").lower()
    
    if feedback == "c":
        print(f"Yay! I guessed your number {guess} in {attempts} attempts.")
        break
    elif feedback == "h":
        high = guess - 1
    elif feedback == "l":
        low = guess + 1
    else:
        print("Invalid input, please enter 'c', 'h', or 'l'.")
        
```
